# 13.6 — Defend DocuMind: the 90-Minute Assessed Defence

Fifteen minutes you drive, sixty minutes of questions across the five published dimensions with your live system on screen, and fifteen minutes of feedback scored there and then. Nothing in the rubric is new on the day &mdash; you have had it since 13.1.


## Cell 1: The Ninety Minutes

And what is actually happening in each of them.


```python
# The ninety minutes, and what is happening in each of them.
BLOCKS = [
    (0,  15, 'You present',
     'architecture, the lane, ONE thing that went wrong. You drive.'),
    (15, 75, 'Interview',
     'two assessors, five dimensions, any order, your live system on screen.'),
    (75, 90, 'Feedback',
     'scored there and then, against the published rubric, reasons written down.'),
]
for a, b, what, detail in BLOCKS:
    print(f'  {a:>2}-{b:<3} {what:14} {detail}')
print()
print('  The first fifteen are yours and most people waste them describing the')
print('  architecture the assessors have already read. Spend them on the thing')
print('  that went wrong: it is the strongest evidence you operated this, and it')
print('  sets up three of the five dimensions.')
print()
print('  "The system is on screen" is not a formality. Every dimension can be')
print('  checked live, and the assessor is told to score the system in front of')
print('  them rather than the story about it.')
print()
DIFFERENT = [
    ('13.5 peer mock',  60, 'a peer',      'unscored', 'practice'),
    ('13.6 defence',    90, 'two assessors', 'scored',  'the credential'),
]
print(f'  {"event":16} {"min":>4} {"with":15} {"scored":10} for')
for ev, m, who, sc, why in DIFFERENT:
    print(f'  {ev:16} {m:>4} {who:15} {sc:10} {why}')
print()
print('  These are two different events and the words are not interchangeable.')
print('  The mock exists so the defence is not the first time you say any of')
print('  this out loud.')
```

## Cell 2: What a Confident Wrong Answer Costs

The assessor says this at the start, so here it is four weeks early.


```python
# What a confident wrong answer costs, and why "I do not know" is cheap.
#
# The assessor is told to say this at the start, so here it is in advance.
ANSWERS = [
    ('"I do not know - I would look at the Cloud Run logs first"',
     'no penalty. It is calibrated, and it names a next step.'),
    ('"About two rupees an answer, I think - I would have to check"',
     'no penalty on Cost. Hedged, honest, and checkable.'),
    ('"Roughly 40 paise" (it is Rs 2.10)',
     'costs you the Cost dimension AND makes the assessor doubt the other four.'),
    ('"We handle that with retries" (there are no retries)',
     'the assessor will ask to see it. This is the expensive one.'),
]
for said, cost in ANSWERS:
    print(f'  {said}')
    print(f'      -> {cost}')
    print()
print('  The rubric rewards CALIBRATION, and it is not being kind. A learner who')
print('  inflates one answer has told the assessor that the other four need')
print('  checking, and there are only sixty minutes.')
print()
print('  Rehearse the not-knowing. It is the hardest part to do well and the')
print('  only part you can practise in a mirror.')
```

## Cell 3: What This Credential Says

And, just as deliberately, what it does not.


```python
# What this credential says, and what it does not.
SAYS = [
    'you shipped a multi-tenant RAG system to a live URL',
    'behind an identity gate you can explain',
    'through a pipeline nobody holds a key to',
    'with an eval suite that has gone red and told you something',
    'and you answered sixty minutes of questions about it, on the system itself',
]
NOT = [
    'that you are a machine-learning researcher',
    'that DocuMind is production-grade for a real customer',
    'that you could build it again from nothing next week',
]
print('  It says:')
for s in SAYS:
    print(f'    - {s}')
print()
print('  It does not say:')
for s in NOT:
    print(f'    - {s}')
print()
print('  That second list is not modesty. It is the reason the first list is')
print('  worth anything: a credential that claimed all three would be a')
print('  credential nobody could check, and this one was designed to be checked')
print('  by a stranger in ten minutes plus ninety.')
print()
print('  Thirteen modules. You started with a project id and a token count.')
```
