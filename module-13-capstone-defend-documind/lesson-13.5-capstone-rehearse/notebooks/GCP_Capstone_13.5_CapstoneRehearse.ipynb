## Cell 1: One Incident, Honestly

Write up the one that surprised you &mdash; not the worst one.


```python
# POST-MORTEM.md. One real incident, four sections, no blame.
#
# You caused four incidents on purpose in 13.3. Write up the one that surprised
# you - not the worst one.
SECTIONS = [
    ('What happened',  'the timeline, in the order YOU learned it - not the true order'),
    ('How it was detected', 'and honestly: did an alert fire, or did you notice?'),
    ('How it was recovered', 'the commands, with timings'),
    ('Action items',   'each with an owner and a date, or it is a wish'),
]
for name, what in SECTIONS:
    print(f'  {name:22} {what}')
print()
print('  Section 2 is the one that earns the mark. "I noticed while looking at')
print('  something else" is a REAL and common answer, and it tells you exactly')
print('  what to build next. Writing "the latency alert fired" when it did not')
print('  is the only wrong answer here.')
print()
print('  And the timeline being in the order YOU learned it is deliberate. A')
print('  reconstructed true timeline hides the twenty minutes you spent looking')
print('  at the wrong service, which is where the useful action item lives.')
```

## Cell 2: Twelve Questions, and What Each Is For

Give this list to your peer. Two of them have no good short answer.


```python
# The mock. Twelve questions your peer should ask, and what each is for.
BANK = [
    ('architecture',   'Draw the path from a browser click to a cited answer.'),
    ('architecture',   'What would make you move off Cloud Run? Give me the number.'),
    ('architecture',   'Why not one project per tenant?'),
    ('failure modes',  'Tell me about a time it broke. Not hypothetically.'),
    ('failure modes',  'Which of your four failure modes would your alerts miss?'),
    ('failure modes',  "What happens if a tenant's chunks disappear from the index right now?"),
    ('cost',           'What does one answer cost, in rupees?'),
    ('cost',           'What breaks first at ten times the volume?'),
    ('cost',           'Which line of the bill would you attack first, and why that one?'),
    ('evals',          'Show me a red run.'),
    ('evals',          'Which of your assertions cannot fail?'),
    ('next month',     'What did operating this teach you that you did not expect?'),
]
by = {}
for dim, q in BANK:
    by.setdefault(dim, []).append(q)
for dim, qs in by.items():
    print(f'  {dim.upper()}')
    for q in qs:
        print(f'      {q}')
    print()
print('  Two of these have no good short answer and that is deliberate:')
print('  "Which of your assertions cannot fail?" and "What did operating this')
print('  teach you that you did not expect?". They are the 3s on their')
print('  dimensions, and they are the two to rehearse.')
print()
print('  Ask them of your peer as well. Hearing an answer that sounds fine and')
print('  realising you cannot check it is how you find the same shape in three')
print('  of your own.')
```

## Cell 3: The Peer Mock Is Not the Defence

Sixty minutes, a peer, unscored &mdash; and it is the highest-value hour in this module.


```python
# The vocabulary one-pagers. Each term hooked to something you built.
#
# The interview asks about these. Answering from a definition is a 1; answering
# from your own system is a 3.
TERMS = [
    ('Mixture of Experts',  '10.6', 'why gpt-oss-20b fine-tunes in 14GB: 4 of 32 experts active'),
    ('Vision Transformer',  '9.6',  'why a chart image costs what it costs in tokens'),
    ('CLIP / SigLIP',       '9.5',  'why image and text land in one vector space'),
    ('Diffusion',           '9.5',  'why generation time scales with steps, not with output size'),
    ('GRPO',                '10.6', 'why reasoning tuning needs a reward, not a label'),
    ('Graph RAG',           '4.6',  'when a join beats a bigger top_k'),
    ('Memory Bank vs Store', '8.6', 'what survives a restart and what does not'),
    ('candidate vs canary',  '12.7', "a no-traffic revision judged before the flip; the canary is the full profile's 10%"),
]
print(f'  {"term":22} {"lesson":8} the hook in YOUR system')
for term, lesson, hook in TERMS:
    print(f'  {term:22} {lesson:8} {hook}')
print()
print('  One page each, and the page must end with a sentence about your fork -')
print('  not about the concept. "MoE means only some experts run" is a')
print('  definition. "We did not use an MoE model; our SLM lane runs a dense 4B')
print('  because our GPU duty cycle is 4% and the memory saving buys nothing at')
print('  that volume" is an answer.')
```
