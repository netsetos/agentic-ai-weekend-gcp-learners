# 13.1 — Kick Off the Capstone: Eight Components, a Published Rubric, and One Lane

You have built DocuMind. The capstone asks you to make it yours, ship it at a live URL through your own pipeline, and answer questions about it for ninety minutes. This lesson publishes the rubric weeks before the defence, names the eight components a stranger will check, and makes you choose one lane and write down what you rejected.


## Cell 1: The Eight Components

Not eight things to talk about. Eight things a stranger can check in ten minutes.


```python
# The eight things that have to exist in your fork.
#
# Not eight things to talk about - eight things a stranger can check in ten
# minutes. That constraint is the whole design: a component nobody can verify
# is a claim, and the defence is sixty minutes of not accepting claims.
COMPONENTS = [
    (1, 'A live URL behind IAP',            'open it, sign in, ask; a non-member gets 403'),
    (2, 'Your own corpus, 30+ pages',       'a query cites one of your documents'),
    (3, 'An eval suite, 30+ rows',          'make eval exits 0; 5+ isolation rows'),
    (4, 'ONE lane on the core',             'chat / media / SLM / graph - chosen, not invented'),
    (5, 'ARCHITECTURE.md + what you rejected', 'three rejected options, with reasons'),
    (6, 'COST-MODEL.md at 1k/10k/100k',     'dual-priced INR, dated, USD_INR = 85'),
    (7, 'FAILURE-MODES.md, 4 entries',      'each proved by a script you can run'),
    (8, 'RUNBOOK.md + a recorded rollback', '/version before and after'),
]
for n, what, check in COMPONENTS:
    print(f'  {n}. {what:38} {check}')
print()
print('  Component 5 is the one people under-build and the one the interview')
print('  leans on hardest. A document listing what you BUILT is a description.')
print('  One listing what you did NOT build, and why, is a decision record -')
print('  and sixty of the ninety minutes are somebody asking why.')
print()
print('  Note what is NOT on the list: a novel feature. The lane is one of four,')
print('  chosen. This credential is about shipping and defending something, not')
print('  about having an idea.')
```

## Cell 2: The Rubric, in Advance

Five dimensions, 0&ndash;3 each, nine to pass &mdash; and one rule that matters more than the total.


```python
# The rubric, as a function. Score yourself before somebody else does.
#
# Five dimensions, 0-3 each, 15 available. NINE to pass - and no dimension at
# zero, which is the rule that matters more than the total.
DIMENSIONS = ['architecture', 'failure_modes', 'cost_at_10x', 'evals', 'next_month']
PASS_TOTAL, MAX_ZERO = 9, 0


def result(scores: dict) -> tuple[bool, str]:
    """Returns (passed, why). Two ways to fail, and they are different."""
    missing = [d for d in DIMENSIONS if d not in scores]
    if missing:
        return False, f'incomplete submission: {missing}'
    zeros = [d for d, s in scores.items() if s == 0]
    total = sum(scores.values())
    if zeros:
        return False, f'zero on {zeros} - total {total}/15 does not rescue it'
    if total < PASS_TOTAL:
        return False, f'{total}/15, below {PASS_TOTAL}'
    return True, f'{total}/15'


CANDIDATES = [
    ('shipped it and can defend every choice',
     {'architecture': 3, 'failure_modes': 3, 'cost_at_10x': 2, 'evals': 3, 'next_month': 2}),
    ('solid build, never operated it',
     {'architecture': 3, 'failure_modes': 0, 'cost_at_10x': 3, 'evals': 3, 'next_month': 2}),
    ('everything works, nothing is explained',
     {'architecture': 1, 'failure_modes': 1, 'cost_at_10x': 1, 'evals': 2, 'next_month': 1}),
    ('rough edges, thought about all of them',
     {'architecture': 2, 'failure_modes': 2, 'cost_at_10x': 2, 'evals': 2, 'next_month': 1}),
]
for label, s in CANDIDATES:
    passed, why = result(s)
    print(f'  {"PASS" if passed else "fail":5}  {label:42} {why}')
print()
print('  Row 2 is the rule doing its job. Eleven of fifteen - comfortably above')
print('  nine - and it fails, because a zero on failure modes means the system')
print('  has never been operated, and no amount of architecture makes up for it.')
print()
print('  Row 4 passes on nine with nothing above a two. That is deliberate: this')
print('  credential is not for the best system in the cohort. It is for somebody')
print('  who can be trusted with one.')
```

## Cell 3: The Section Everybody Under-Builds

ARCHITECTURE.md is not a description of what you built. It is a list of what you did not.


```python
# ARCHITECTURE.md is not a description. It is a list of things you did not do.
#
# The rubric's 3 for architecture is a FALSIFIABLE statement - a threshold, not
# a feeling. Here is the shape.
DECISIONS = [
    ('hosting',
     'Cloud Run',
     'GKE Autopilot',
     'our duty cycle is 4%; GKE wins above ~74% (11.5). We would move at 26,000 answers/day.'),
    ('retrieval',
     'Firestore vectors alone (PROFILE=lean)',
     'Vector Search + a Firestore fallback (PROFILE=full)',
     '1,624 chunks per tenant and the 64-row gate clears at 97.8% on Firestore alone. Past ~40k chunks its p95 goes from 180ms to ~1.9s (13.3); that is where we would flip the profile.'),
    ('model',
     'gemini-3.6-flash',
     'a self-hosted Gemma (11.4)',
     'self-hosting is cheaper above ~74% GPU duty cycle; we are at 4%. And one L4 is all the lane allows itself: the quota is capped at 1 (make gpu-cap), and Mumbai has none to give.'),
    ('tenancy',
     'one project, tenant_id predicate',
     'a project per tenant',
     'we have 3 tenants. At ~50 the per-project quota work costs more than the isolation is worth.'),
]
for area, chose, rejected, why in DECISIONS:
    print(f'  {area:12} chose {chose}')
    print(f'  {"":12} over  {rejected}')
    print(f'  {"":12}       {why}')
    print()
print('  Every "why" above contains a NUMBER and a condition that would reverse')
print('  the decision. That is the difference between a 2 and a 3.')
print()
print('  "We chose Cloud Run because it is serverless" is a 1. It describes the')
print('  product, not the choice - it would be equally true if you had chosen')
print('  wrong.')
```
