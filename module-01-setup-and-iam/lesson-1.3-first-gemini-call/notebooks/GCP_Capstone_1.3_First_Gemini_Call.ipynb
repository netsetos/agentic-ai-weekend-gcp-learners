## Cell 1: Setup


```python
!pip install -q google-genai==2.21.0
from google.colab import auth
auth.authenticate_user()

PROJECT_ID = 'documind-ai-YOUR-ID'  # CHANGE THIS
# Gemini 3.x generation is served only from the global endpoint; a regional
# value like 'us-central1' returns 404. (Regional services still use a region.)
LOCATION = 'global'
USD_TO_INR = 85
```

## Cell 2: Initialize Client & First Call


```python
from google import genai
from google.genai import types

client = genai.Client(enterprise=True, project=PROJECT_ID, location=LOCATION)

response = client.models.generate_content(
    model='gemini-3.6-flash',
    contents='What is a token in LLM terminology? Answer in 2 sentences.'
)
print('Response:', response.text)
```

## Cell 3: Decode usage_metadata


```python
meta = response.usage_metadata
print('Token Breakdown:')
print(f'  Input tokens:    {meta.prompt_token_count}')
print(f'  Output tokens:   {meta.candidates_token_count}')
print(f'  Thinking tokens: {(getattr(meta, "thoughts_token_count", 0) or 0)}')
print(f'  Total tokens:    {meta.total_token_count}')
```

## Cell 4: Calculate Cost in INR


```python
# STANDARD rates, USD per 1M tokens (apply from 2027-01-01)
PRICING = {
    'gemini-3.1-flash-lite': {'input': 0.25, 'output': 1.50},
    'gemini-3.6-flash':      {'input': 1.50, 'output': 7.50},
    'gemini-3.1-pro-preview':        {'input': 2.00, 'output': 12.00},
}
# INTRODUCTORY rate, 3.6 Flash only - valid through 2026-12-31, then the standard rate above applies
PRICING_INTRO = {
    'gemini-3.6-flash':      {'input': 0.75, 'output': 3.75},
}

def price_for(model, rate='intro'):
    """USD per 1M tokens. rate='intro' = 3.6 Flash introductory price (to 2026-12-31);
    rate='standard' = rates from 2027-01-01. Flash-Lite and Pro are identical on both."""
    if rate == 'intro' and model in PRICING_INTRO:
        return PRICING_INTRO[model]
    return PRICING[model]

def calc_cost_inr(meta, model='gemini-3.6-flash', rate='intro'):
    """Cost of one call in INR. rate='intro' (default while it is valid, to 2026-12-31)
    or 'standard' (from 2027-01-01). Thinking tokens are billed at the output rate."""
    p = price_for(model, rate)
    inp = meta.prompt_token_count * p['input'] / 1_000_000
    think = (getattr(meta, 'thoughts_token_count', 0) or 0)
    out = (meta.candidates_token_count + think) * p['output'] / 1_000_000
    total = (inp + out) * USD_TO_INR
    print(f'  [{rate} rate] Input: Rs.{inp*USD_TO_INR:.4f} | Output+Think: Rs.{out*USD_TO_INR:.4f} | Total: Rs.{total:.4f}')
    print(f'  With $500: {500/(inp+out):,.0f} calls possible')
    return total

calc_cost_inr(response.usage_metadata)
```

## Cell 5: Free Token Counter


```python
texts = [
    'Hello',
    'Hyderabad is the capital of Telangana' * 10,
    'The transformer architecture uses self-attention' * 100,
]

for t in texts:
    r = client.models.count_tokens(model='gemini-3.6-flash', contents=t)
    print(f'  {r.total_tokens:>6,} tokens | {len(t):>6} chars | {len(t)/r.total_tokens:.1f} chars/token | {t[:40]}...')
```

## Cell 6: Compare 3 Models


```python
import time

prompt = 'Explain the difference between SQL and NoSQL. Give 2 examples of each.'
models = ['gemini-3.1-flash-lite', 'gemini-3.6-flash', 'gemini-3.1-pro-preview']

print(f"{'Model':<35} {'Tokens':>7} {'Latency':>9} {'Rs':>9}")
print('-' * 65)
for name in models:
    t0 = time.time()
    try:
        r = client.models.generate_content(model=name, contents=prompt)
        ms = (time.time()-t0)*1000
        u = r.usage_metadata
        p = price_for(name)   # Cell 4 helper: intro rate for 3.6 Flash (to 2026-12-31); thinking billed at output rate
        cost = (u.prompt_token_count*p['input'] + (u.candidates_token_count + (getattr(u, 'thoughts_token_count', 0) or 0))*p['output'])/1e6*USD_TO_INR
        print(f'{name:<35} {u.total_token_count:>7} {ms:>7.0f}ms Rs.{cost:>6.4f}')
    except Exception as e:
        print(f'{name:<35} ERROR: {e}')
```

## Cell 7: Thinking Budget Impact


```python
prompt = 'What are the pros and cons of microservices?'

# Gemini 3.x thinks at a LEVEL, never off; LOW is the floor and what the kit's API runs at.
print(f"{'Level':>8} {'Think':>7} {'Output':>8} {'Total':>7} {'Rs.':>9}")
for b in ['LOW', 'HIGH']:
    cfg = types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(thinking_level=b),
        # Safety: block only HIGH-probability harassment (Vertex default for Gemini 3.x is OFF, so this is stricter). Note: no temperature/top_p/top_k -
        # gemini-3.6-flash ignores them, and frequency/presence penalties are rejected.
        safety_settings=[types.SafetySetting(category='HARM_CATEGORY_HARASSMENT', threshold='BLOCK_ONLY_HIGH')],
    )
    r = client.models.generate_content(model='gemini-3.6-flash', contents=prompt, config=cfg)
    u = r.usage_metadata
    think = (getattr(u, 'thoughts_token_count', 0) or 0)
    p = price_for('gemini-3.6-flash')   # intro rate to 2026-12-31 (Cell 4 helper)
    cost = (u.prompt_token_count*p['input'] + (u.candidates_token_count+think)*p['output'])/1e6*USD_TO_INR
    print(f'{b:>8} {think:>7} {u.candidates_token_count:>8} {u.total_token_count:>7} Rs.{cost:>7.4f}')
```

## Cell 8: Streaming


```python
print('Streaming response:')
last = None
for chunk in client.models.generate_content_stream(
    model='gemini-3.6-flash',
    contents='Write a haiku about machine learning.'
):
    print(chunk.text or '', end='', flush=True)   # thought / empty chunks carry text=None
    last = chunk
# usage_metadata is on the FINAL chunk only
usage = last.usage_metadata if last else None
total = usage.total_token_count if usage else 'n/a'
print(f'\n\nTotal tokens: {total}')
```

## Cell 9: Monthly Cost Projector


```python
def project_monthly(qpd, avg_in, avg_out, mix={'flash-lite':0.6,'flash':0.3,'pro':0.1}):
    # STANDARD rates (from 2027-01-01): a multi-year projection uses the rate in force for most of
    # the horizon. The 3.6 Flash intro rate ($0.75/$3.75) only runs to 2026-12-31.
    prices = {'flash-lite':(0.25,1.50), 'flash':(1.50,7.50), 'pro':(2.00,12.00)}
    monthly = qpd * 30
    total = 0
    print(f'Queries/day: {qpd} | Monthly: {monthly:,} (standard 2027 rates)')
    for tier, pct in mix.items():
        ip, op = prices[tier]
        q = monthly * pct
        cost = (q*avg_in*ip + q*avg_out*op)/1e6
        total += cost
        print(f'  {tier:<12} {pct*100:>4.0f}% | {q:>7,.0f}q | ${cost:>6.2f} | Rs.{cost*85:>8.2f}')
    print(f'\n  TOTAL: ${total:.2f}/mo (Rs.{total*85:.2f})')
    print(f'  $500 lasts: {500/total:.1f} months ({500/total/12:.1f} years) at standard 2027 rates')

project_monthly(100, 2000, 500)
```
